Question 1 : How do the average scores, total kills, and total deaths vary across different player levels and regions? (3D & 2D Scatter Plot)

```python
# 3D Scatterplot

import plotly.express as px
import pandas as pd

# Load the dataset
df = pd.read_csv('/Users/corydangus/Downloads/fortnite_players_stats.csv')

# Check the dataset columns to map the actual names to the intended data
print(df.columns)

# Replace the column names in scatter_3d with correct names from the dataset
# Assuming:
# - 'Solo matches' is the player level (number of matches represents experience)
# - 'Solo kills' is the total kills
# - 'Solo minutesPlayed' is the total deaths proxy (assuming higher playtime correlates to more deaths)
# - 'Solo winRatio' is used as the region (color grouping), and
# - 'Solo score' as the score size indicator.

fig = px.scatter_3d(
    df,
    x='Solo matches',            # Replacing 'Player Level'
    y='Solo kills',              # Replacing 'Total Kills'
    z='Solo minutesPlayed',      # Replacing 'Total Deaths'
    color='Solo winRatio',       # Replacing 'Region'
    size='Solo score',           # Replacing 'Score'
    size_max=20
)

fig.update_layout(
    title='3D Scatter: Player Performance by Matches, Kills, and Playtime',
    scene=dict(
        xaxis_title='Matches Played',
        yaxis_title='Kills',
        zaxis_title='Minutes Played'
    )
)

fig.show()
```

```python
import plotly.express as px
import pandas as pd

# Load the dataset
df = pd.read_csv('/Users/corydangus/Downloads/fortnite_players_stats.csv')

# Select relevant columns for analysis
# Assume: Solo matches = Player experience, Solo kills = Performance, Solo score = Overall success metric
subset = df[['Solo matches', 'Solo kills', 'Solo score', 'Solo winRatio']]

# Create a scatter plot matrix
fig = px.scatter_matrix(
    subset,
    dimensions=['Solo matches', 'Solo kills', 'Solo score', 'Solo winRatio'],
    color='Solo winRatio',  # Use win ratio as color for better insights
    title='Scatter Matrix: Fortnite Player Performance Metrics',
    labels={
        'Solo matches': 'Matches Played',
        'Solo kills': 'Kills',
        'Solo score': 'Score',
        'Solo winRatio': 'Win Ratio (%)'
    },
    template='plotly_white'
)

fig.update_layout(
    title_font_size=16,
    xaxis_title_font_size=12,
    yaxis_title_font_size=12
)

fig.show()
```
